```python
# Step 1 - Read Twitter Avro dataset

tweets_df = spark.read.format("avro").load(
    "/Volumes/workspace/retail_lab/raw_files/tweets.avro"
)

display(tweets_df)
```

```python
# Step 2 - Inspect the Avro schema

tweets_df.printSchema()
```

```python
# Question 1 - Top 5 users with the highest number of tweets

display(
    tweets_df.groupBy("user.username")
    .count()
    .orderBy("count", ascending=False)
    .limit(5)
)
```

```python
from pyspark.sql import functions as F
```

```python
# Question 2 - Split tweets into words

words_df = tweets_df.select(
    F.split(F.lower(F.col("text")), " ").alias("words")
)

display(words_df)
```

```python
# Remove English stopwords

from pyspark.ml.feature import StopWordsRemover

remover = StopWordsRemover(
    inputCol="words",
    outputCol="filtered_words"
)
```

```python
# Prepare tweets as lists of words

words_df = tweets_df.select(
    F.split(F.lower(F.col("text")), " ").alias("words")
)

display(words_df)
```

```python
# Remove English stopwords

words_df = remover.transform(words_df)

display(words_df)
```

```python
# Count the most common words

display(
    words_df.select(F.explode("filtered_words").alias("word"))
    .filter(~F.col("word").startswith("@"))
    .filter(~F.col("word").startswith("#"))
    .withColumn("word", F.regexp_replace("word", r"[^a-z]", ""))
    .filter(F.col("word") != "")
    .groupBy("word")
    .count()
    .orderBy("count", ascending=False)
    .limit(10)
)
```

```python
# Clean tweet text before removing stopwords

words_df = tweets_df.select(
    F.split(
        F.lower(F.regexp_replace("text", r"[^a-zA-Z\s]", "")),
        " "
    ).alias("words")
)

remover = StopWordsRemover(
    inputCol="words",
    outputCol="filtered_words",
    stopWords=StopWordsRemover.loadDefaultStopWords("english") + ["rt"]
)

words_df = remover.transform(words_df)

display(words_df)
```

```python
# Question 2 - Find the 10 most common words

display(
    words_df.select(F.explode("filtered_words").alias("word"))
    .groupBy("word")
    .count()
    .orderBy("count", ascending=False)
    .limit(10)
)
```

```python
# Question 3 - Filter tweets in Spanish

display(
    tweets_df.filter(F.col("lang") == "es")
)
```

```python
# Check the languages available in the dataset

display(
    tweets_df.groupBy("lang")
    .count()
    .orderBy("count", ascending=False)
)
```
